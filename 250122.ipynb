# 3, 6, 9가 있을 때 마다 박수를 치는 3,6,9 게임을 만들어봅시다.

## 1) 사용자에게 입력받아 원하는 길이만큼 3,6,9 게임을 혼자 실행하는 코드

## 2) 컴퓨터와 같이 3,6,9를 하면서, 지금 차례의 숫자 혹은 박수를 입력하기 (틀리면 패배)

```python
my_turn = True
for i in range(1, 100):
  if my_turn == True:
    user = input("당신 차례입니다: (3,6,9면 짝)")
    #사용자가 3,6,9라고 판단한 경우
    if user == "짝":
      # if i가, 3 6, 9인가?
      i = str(i)
      for j in range(len(i)):
        # 3,6,9면
        if i[j] == "3" or i[j] == "6" or i[j] == "9":
          break
        else:
          print("패배")

    #사용자가 3,6,9가 아니라고 판단한 경우
    else:
      i = str(i)
      for j in range(len(i)):
      # 3,6,9가 맞는데, 아니라고 판단.
        if i[j] == "3" or i[j] == "6" or i[j] == "9":
          print("패배")
          break

    my_turn = False

  # 컴퓨터 차례
  else:
    i = str(i)
    박수 = False

    for j in range(len(i)):
      if i[j] == "3" or i[j] == "6" or i[j] == "9":
        print("컴퓨터 : ", "짝")
        박수 = True
        break

      else:
        박수 = False
        break
    my_turn = True
```

```python
def user_input():
    return input("사용자 차례: ")

def check_369(number):
  number = str(number)
  clap_count = 0
  for i in range(len(number)):
    if number[i] == "3" or number[i] == "6" or number[i] == "9":
      clap_count += 1

  return clap_count

def check_correct(now_number, answer):
  clap = check_369(now_number)
  if clap:
    if answer == '짝'*clap:
      print("")
      return True

    else:
      return False
  else:
    return now_number == answer

def change_computer_answer(number):
  clab = check_369(number)
  if clab:
    return "짝"*clab
  else:
    return number

print("시작")
user_turn = True
for now_number in range(1,100):
  now_number = str(now_number)
  if user_turn:
    user_answer = user_input()
    if not check_correct(now_number, user_answer):
      print("패배")
      break
    user_turn = False
  else:
    computer_answer = change_computer_answer(now_number)
    print("컴퓨터 차례:", computer_answer)
    user_turn = True
```

```python
def check_now_369(number):
  number = str(number)
  clap = False
  for i in range(len(number)):
    if number[i] == "3" or number[i] == "6" or number[i] == "9":
      clap = True

  return clap


def check_correct(number, input_value):
  if check_now_369(number):
    if input_value == "짝":
      return True

    else:
      return False

  else:
    return number == input_value

for now_number in range(1,100):
  if user_turn:
    value = user_input()
    if check_correct(now_number, value):
      user_turn = False

    else:
      break

  else:
    now_number = change_computer_answer(now_number)
    print(now_number)
    user_turn = True
```

## 단어가 주어지고, 글자를 맞출 수 있는 기회가 주어짐

## 기회 횟수가 끝날 때 까지 못하면 실패
## 이전에 맞추면 성공

```python
word = "파이파"
print(f"글자 수는 {len(word) }입니다.")
word_status = "_" * len(word)
print(f"현재 상태: {word_status}")
chance = 5
print("남은 기회: ", chance)

while chance > 0:
  print("현재상태",word_status)
  user_answer = input("글자를 입력해주세요: ")

  index = word.find(user_answer)
  # 맞춘 경우
  if index != -1:
    new_answer = ""
    for i in range(len(word)):

      if i != index:
        new_answer += word_status[i]
      else:
        new_answer += user_answer

    word_status = new_answer
    print("새로운상태",word_status)

    if word_status == word:
      print("성공!")
      chance = 0

  else:
    chance -= 1
    print("오답입니다! 남은 기회 :", chance)


```
